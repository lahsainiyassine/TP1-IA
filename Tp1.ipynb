```python
# 1. Installation des packages (à exécuter dans le terminal ou avec '!' dans un notebook)
# !pip install nltk spacy scikit-learn

# 2. Téléchargement des ressources linguistiques
import nltk
import spacy

# Téléchargement des modules NLTK nécessaires
nltk.download('punkt')          # Pour la tokenisation
nltk.download('stopwords')      # Pour les listes de mots vides
nltk.download('wordnet')        # Pour la lemmatisation (anglais)
nltk.download('omw-1.4')

# Pour spaCy : télécharger le modèle français (dans le terminal : python -m spacy download fr_core_news_sm)
nlp = spacy.load("fr_core_news_sm")
```

```python
import string
import re

# Phrase brute de test
text = "L'Intelligence Artificielle (IA) transforme le monde moderne... Mais est-elle toujours fiable à 100% ?!"

# 1. Passage en minuscules
text_lower = text.lower()

# 2. Suppression de la ponctuation et caractères spéciaux
# Méthode avec regex pour ne conserver que les lettres, chiffres et espaces
text_cleaned = re.sub(r'[^\w\s]', ' ', text_lower)
# Suppression des espaces multiples consécutifs
text_cleaned = re.sub(r'\s+', ' ', text_cleaned).strip()

print("Avant :", text)
print("Après :", text_cleaned)
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# 1. Tokenisation avec NLTK
tokens = word_tokenize(text_cleaned, language='french')
print("Tokens originaux :", tokens)

# 2. Récupération des stopwords en français
stop_words = set(stopwords.words('french'))

# 3. Ajout d'un mot personnalisé qui n'apporte pas de sens au corpus (ex: 'ia')
custom_stopword = 'ia'
stop_words.add(custom_stopword)

# 4. Filtrage des tokens
filtered_tokens = [w for w in tokens if w not in stop_words and len(w) > 1]
print("Tokens filtrés :", filtered_tokens)

# 5. Calcul de la proportion supprimée
nb_init = len(tokens)
nb_filtres = len(filtered_tokens)
proportion_supprimee = ((nb_init - nb_filtres) / nb_init) * 100

print(f"Proportion supprimée : {proportion_supprimee:.2f}%")
```

```python
# spaCy gère l'analyse morpho-syntaxique complète (POS) nécessaire à la lemmatisation en français
doc = nlp(" ".join(filtered_tokens))

# Comparaison avant et après lemmatisation
lemmatized_results = [(token.text, token.lemma_) for token in doc]

print(f"{'Token filtré':<20} | {'Lemme canonique'}")
print("-" * 42)
for original, lemma in lemmatized_results:
    print(f"{original:<20} | {lemma}")
```
